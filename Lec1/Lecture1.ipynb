Truth table for <code>and</code> and <code>or</code>

```python
print("  a   |    b  || a and b | a or b")
print("------+-------++---------+-------")
print("False | False ||  {}  |  {}".format(False and False, False or False))
print("False | True  ||  {}  |  {}".format(False and True, False or True))
print("True  | False ||  {}  |  {}".format(True and False, True or False))
print("True  | True  ||  {}   |  {}".format(True or True, True or True))
```

Order or operations is important. We have <code>not</code> before <code>and</code> and <code>or</code>:

```python
a = True
b = False
print("not a and b   =", not a and b)
print("not (a and b) =", not (a and b))
print("not a or b    =", not a or b)  # "not" before "or"!
```

**Note:** The <code>and</code> and <code>or</code> operators are "intelligent" / "anticipatory":
* <code>A and B</code>: If <code>A</code> is false, then <code>B</code> will not be evaluated
* <code>A or B</code>: If <code>A</code> is true, then <code>B</code> nicht evaluiert

This is useful for more complex expressions but dangerous, as the following example shows:

```python
some_variable = True
F = some_variable or IAmNotDefined
print(F)
some_variable = False
F = some_variable and I(Am(Not(Defined(2 * nothing))))
print(F)
```

 * <span style="font-weight: bold; color:red">ATTENTION</span>: The above code does not give an error even though, for example, <code>IAmNotDefined</code> is not defined.
 * Compiled programs would report an error here when compiling, because the variable or expression <code>IAmNotDefined</code> was not defined anywhere!
 * This can lead to errors that are very difficult to identify. Especially when very complex operations are necessary to determine some_variable.

```python
some_variable = True
F = some_variable or IAmNotDefined
print(F)
#If we change 'some_variable' to False, it yields an error
some_variable = False
F = some_variable or IAmNotDefined
print(F)
```

```python
some_variable = False
IAmDefined = 'Hello, I am defined!'

F = some_variable or IAmDefined
print(F)
```

If the operands involved in an <code>or</code> operation are objects instead of Boolean expressions, then the Python <code>or</code> operator returns a true or false object, not the values <code>True</code> or <code>False</code> as you could expect. 

By default, an object is considered true unless its class defines either a <code>\_\_bool\_\_()</code> method that returns <code>False</code> or a <code>\_\_len\_\_()</code> method that returns zero, when called with the object. Here are most of the built-in objects considered false (https://docs.python.org/3/library/stdtypes.html#truth-value-testing): 

 * constants defined to be false: None and False

 * zero of any numeric type: 0, 0.0, 0j, Decimal(0), Fraction(0, 1)

 * empty sequences and collections: '', (), [], {}, set(), range(0)

## The <code>while</code>-loop

Often certain instructions have to be executed repeatedly. Loops and other control structures exist for this purpose.

```python
i = 1
# "As long as" i <= 10
while i <= 10:
    print(i)
    i = i + 1
```

* The <code>while</code> loop "runs" till the condition <code>i &lt;= 10</code> holds, i.e. it terminates as soon as $i > 10$ is "true".
* After the colon, a so-called **block** begins, which is defined by indenting the source text with 4 spaces.
* The <code>while</code> loop executes all commands within the block.

```python
i = 1
while i <= 10:
    # Compute the square and root of the number i
    square = i*i
    sqrt = i**(1/2)

    print("{:2d} | {:3d} | {:.4f}".format(i, square, sqrt))
    i = i + 1
    
# Continue outside the while-block
print("These are the squares and roots of the numbers from 1 to 10.")
```

## About blocks in Python

* **IMPORTANT**: Correct indentation is important here! The complete block must be indented with an identical number of spaces.
* 4 spaces are recommended and provide good readability of the code.
* The 4 spaces are only a convention! For example, only 2 spaces can be used.
* On the other hand, it is not necessary to indent empty lines.
* An IDE/an intelligent editor is very helpful here!

```python
a = 1
while a < 10:
      print("widely indented")

      a = a + 5

while a > 1:
 print("Not so widely indented")
 a = a - 5
```

## The `for`-loop

```python
# Find all divisors of the numbers 1 to number
# BLOCK 1
number = 19
for i in range(1, number): 
    # BLOCK 2
    print("Investigate number", i)
    
    # We don't want to know that every number is divisible by 1:
    for j in range(2, i):
        # BLOCK 3
        if i % j == 0:
            print("  - divisible by", j)
    
    # BLOCK 2 - continuation
    # Blank line for better readability
    print()
    
# BLOCK 1 - continuation
print("The program is ready!")
```

The `range()` function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and stops before a specified number.

```python
x = range(2, 10, 3)
for n in x:
    print(n) 
```

`for` loops also have an `else` clause. The `else` clause executes after the loop completes normally. This means that the loop did not encounter a `break` statement. 

```python
max_attempts = 3
for i in range(max_attempts):
    number = float(input("Insert a number between 1 and 10: "))
    if number > 1 and number < 10:
        break
    print("Try again!")
else:
    print('You have tried {} times. You have no more attempts.'.format(max_attempts))
print("The number", number, "is between 1 and 10!")
```

### Extract partial lists

Python provides built-in mechanisms to extract partial lists. **Important**: For all index operations the start index is always added, but the last index is ignored!

```python
s = [1, 44, ['s', 'p'], '4', 5]
print("s[0:2] =", s[0:2])
print("s[0:len(s)] =", s[0:len(s)])
```

```python
# If the second argument is omitted, the program always iterates to the end:
print("s[0:] =", s[0:])  # Index 0,1,2,3,4
print("s[:3] =", s[:3])  # Index 0,1,2
```

```python
# We extract from the list in two-steps:
print("s[0:len(s):2] =", s[0:len(s):2])
# And the matching abbreviated spelling:
print("s[::2] =", s[::2])
```

The elements of the list can also be indexed starting from the end:

```python
s = [1, 44, ['s', 'p'], '4', 5]
# last element:
print(s[-1])
# last 3 elements:
print(s[-3:])
# 2-nd to the penultimate element:
print(s[1:-1])
```

### List comprehension

```python
base_list = [5,9,22,38]

new_list = []
for l in base_list:
    new_list.append(l**2+1)

print(new_list)
```

This can be done in a much more comprehensive way:

```python
base_list = [5,9,22,38]

new_list = [l**2+1 for l in base_list]

print(new_list)
```

Optional one can add an `if` statement:

```python
other_list=[l+1 for l in base_list if l%2 == 0 ]

print(other_list)
```

```python
A = {1, 2, 3}
B = {1, 2, 3, 4, 5, 2}
print("B = ", B)
x = 2
print("2 element of A: ", x in A)
print("6 element of B: ", 6 in B)
```

```python
# A union B:
print("A ∪ B =", A.union(B))
print("A ∪ B =", A | B)
# A intersection B
print("A ∩ B =", A.intersection(B))
print("A ∩ B =", A & B)
# relative complement (A without B)
print("A ∖ B =", A.difference(B))
print("A ∖ B =", A - B, " (corresponds to empty set)")
print("B ∖ A =", B - A)
```

# Mutable / Immutable

```python
# What does the following code output?
my_list = [1, 2, 3, 4]
second_list = my_list
second_list[1] = 0
print(second_list)
print(my_list)
```

So with the second also the first list is modified!!!

We analyze the above code again more exactly, by looking at the memory addresses of the variables (by means of the `id()` command):

```python
my_list = [1, 2, 3, 4]
print("Address of my_list:", id(my_list))
second_list = my_list 
print("Address of second_list:", id(second_list))
second_list[1] = 0
print("Address of my_list:", id(my_list))
print("Address of second_list:", id(second_list))
```

Apparently all variables contain the same object, in this case the same list.

One calls `second_list` in the above example a "reference" to the first variable, so practically another name for the same object.

We defere between **mutable** (changeable) and **immutable** (not changeable) data types.

**Mutable** data types can change their content without changing their identity (memory address).
Examples:
  * Lists: <code>list</code>
  * Dictionaries: <code>dict</code>
  * Sets: <code>set</code>
  
  
**Immutable** data types cannot be changes.
Examples:
  * Numbers: <code>int, float, complex</code>
  * Bools: <code>bool</code>
  * Tuples: <code>tuple</code>
  * Strings: <code>str</code>
  * and a few more: <code>decimal, frozenset, bytes, range</code>

```python
a = 5
b = a
print("id von a ist {}, id von b ist {}".format(id(a),id(b)))
b = 3
print("id von a ist {}, id von b ist {}".format(id(a),id(b)))
```

### Arbitrary arguments 

If you do not know how many arguments that will be passed into your function, add a `*` before the parameter name in the function definition.

This way the function will receive a tuple of arguments, and can access the items accordingly:

```python
def my_function(*kids):
  print("The youngest child is " + kids[2])

my_function("Emil", "Tobias", "Linus") 
```

### Arbitrary Keyword Arguments, `**kwargs`

If you do not know how many keyword arguments that will be passed into your function, add two asterisk: `**` before the parameter name in the function definition.

This way the function will receive a dictionary of arguments, and can access the items accordingly:

```python
def my_function(**kid):
    print(type(kid))
    print(kid)
    print("His last name is " + kid["lname"])

my_function(fname = "Tobias", lname = "Refsnes") 
```

Mutable data type as key-argument

```python
def append_or_create_list(value, my_list=[]):
    my_list.append(value)
    return my_list
print(append_or_create_list(5))
print(append_or_create_list(10))
```

Let us analyse what is happening here:

```python
def append_or_create_list(value, my_list=[]):
    my_list.append(value)
    return my_list
list1 = append_or_create_list(5)
print("id(list1) =", id(list1))
list2 = append_or_create_list(10)
print("id(list2) =", id(list2))
```

* I.e. the addresses (and thus the lists) are identical.
* The reason is that the function arguments are evaluated and stored once when a function is defined.
* How can we now fix the above code?

```python
def append_or_create_list(value, my_list=None):
    # If my_list is None, then we create a new list:
    if my_list is None:
        my_list = []
    my_list.append(value)
    return my_list

list1 = append_or_create_list(1)
print(list1)
list2 = append_or_create_list(2)
print(list2)
```

## Documentation for functions

Comments are not only important in the program text. Especially for functions a help should ALWAYS be provided. In Python a help can be stored as follows:

```python
def my_well_documented_function(arg1, arg2=5):
    "The function computes the sum of arg1 and arg2"
    return arg1 + arg2

help(my_well_documented_function)
```

Python evaluates the string (without variable assignment) after the definition of the function as a so-called doc-string. You can query the doc-string of a function as follows:

```python
print(my_well_documented_function.__doc__)
```

```python
# An example from the scipy.optimize module.
import scipy.optimize as so
print(so.fsolve.__doc__)
```

### Function documentation

* We **always** create documentation for each function!
* The description should contain a short summary of the function.
* If the arguments are more complex, they should be explained!
* In exceptional cases, documentation can be omitted:
    * Auxiliary functions
    * Absolutely trivial functions
